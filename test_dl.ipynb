```python
url = "https://olmo-data.org/dolma-v1_7/wiki/wiki-0000.json.gz"

```

```python
import gzip, io, json, urllib.request

N = 1000

req = urllib.request.Request(url, headers={"User-Agent": "curl/8.0"})
with urllib.request.urlopen(req) as resp:
    with gzip.GzipFile(fileobj=resp) as gz:
        lines = [json.loads(line) for line, _ in zip(io.TextIOWrapper(gz, encoding="utf-8"), range(N))]

print(len(lines))
```

```python
import shutil
from pathlib import Path

path = Path("data") / url.rsplit("/", 1)[1]
path.parent.mkdir(exist_ok=True)

req = urllib.request.Request(url, headers={"User-Agent": "curl/8.0"})
with urllib.request.urlopen(req) as resp, path.open("wb") as f:
    shutil.copyfileobj(resp, f)

print(path, f"{path.stat().st_size / 1e6:.0f} MB")
```

```python
raw = path.with_suffix("")  # data/wiki-0000.json

with gzip.open(path, "rb") as src, raw.open("wb") as dst:
    shutil.copyfileobj(src, dst)

# byte offset where each line starts
offsets = []
with raw.open("rb") as f:
    pos = 0
    for l in f:
        offsets.append(pos)
        pos += len(l)

def line(k):
    """Returns the k-th (0-based) JSON document, by seeking to its byte offset."""
    with raw.open("rb") as f:
        f.seek(offsets[k])
        return json.loads(f.readline())

print(f"{raw} {raw.stat().st_size / 1e9:.1f} GB, {len(offsets):,} lines")
line(4258)
```
